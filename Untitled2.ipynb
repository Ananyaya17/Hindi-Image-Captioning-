```python
# ============================================================
# CELL 4 — Load Captions
# ============================================================

image_path = '/content/dataset/Images_all'
caption_file = '/content/dataset/captions.txt'

captions = {}

with open(caption_file, 'r', encoding='utf-8') as f:
    lines = f.readlines()

for line in lines:
    img, cap = line.strip().split(',', 1)
    img = img + ".jpg"

    cap = "startseq " + cap + " endseq"

    if img not in captions:
        captions[img] = []
    captions[img].append(cap)

print("Total images:", len(captions))
```

```python
# ============================================================
# CELL 5 — Split Dataset
# ============================================================

from sklearn.model_selection import train_test_split

images = list(captions.keys())

train, temp = train_test_split(images, test_size=0.3, random_state=42)
val, test = train_test_split(temp, test_size=0.5, random_state=42)

print("Train:", len(train))
print("Val:", len(val))
print("Test:", len(test))
```

```python
# ============================================================
# CELL 6 — Tokenizer
# ============================================================

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

all_captions = []

for key in captions:
    all_captions.extend(captions[key])

tokenizer = Tokenizer(oov_token="<unk>")
tokenizer.fit_on_texts(all_captions)

vocab_size = len(tokenizer.word_index) + 1

max_length = max(len(c.split()) for c in all_captions)

print("Vocab size:", vocab_size)
print("Max length:", max_length)
```

```python
features = np.load(
    "/content/drive/MyDrive/spatial_features.npy",
    allow_pickle=True
).item()

print("✅ Features loaded:", len(features))
```

```python
# ============================================================
# CELL 8 — BUILD FINAL MODEL
# ============================================================

from tensorflow.keras.layers import *
from tensorflow.keras.models import Model

class TopKRegions(tf.keras.layers.Layer):

    def __init__(self, k=8):
        super().__init__()
        self.k = k

    def call(self, x):
        # Simply take first K regions (stable + no gradient issue)
        return x[:, :self.k, :]


class BahdanauAttention(tf.keras.layers.Layer):
    def __init__(self, units, **kwargs):
        super().__init__(**kwargs)
        self.W1 = Dense(units)
        self.W2 = Dense(units)
        self.V  = Dense(1)

    def call(self, features, hidden):
        hidden = tf.expand_dims(hidden, 1)
        score = tf.nn.tanh(self.W1(features) + self.W2(hidden))
        weights = tf.nn.softmax(self.V(score), axis=1)
        context = tf.reduce_sum(weights * features, axis=1)
        return context

    def get_config(self):
        config = super().get_config()
        config.update({"units": 512})
        return config


def build_model():

    img_input = Input(shape=(64,2048))
    x = Dense(512, activation='relu')(img_input)

    att = MultiHeadAttention(num_heads=4, key_dim=64)(x, x)
    x = Add()([x, att])
    x = LayerNormalization()(x)

    top_regions = TopKRegions()(x)

    txt_input = Input(shape=(max_length,))
    t = Embedding(vocab_size, 256)(txt_input)

    t = Bidirectional(LSTM(256))(t)
    t = Dense(512, activation='relu')(t)

    attention_global = BahdanauAttention(512)
    attention_region = BahdanauAttention(512)

    context1 = attention_global(x, t)
    context2 = attention_region(top_regions, t)

    merged = Add()([context1, context2])
    merged = LayerNormalization()(merged)

    merged = Dense(512, activation='relu')(merged)
    merged = Dropout(0.4)(merged)

    output = Dense(vocab_size, activation='softmax', dtype='float32')(merged)

    model = Model([img_input, txt_input], output)

    return model


model = build_model()

model.compile(

    loss=tf.keras.losses.CategoricalCrossentropy(
        label_smoothing=0.1,
        from_logits=False
    ),

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4,
        clipnorm=0.5
    ),

    metrics=['accuracy']
)

model.summary()
```

```python
# ============================================================
# CELL 9 — DATA GENERATOR (FIXED + OPTIMIZED)
# ============================================================

def data_generator(image_list, batch_size):

    while True:

        X1, X2, y = [], [], []

        for img in image_list:

            if img not in captions or img not in features:
                continue

            feature = features[img]

            for cap in captions[img]:

                seq = tokenizer.texts_to_sequences([cap])[0]

                for i in range(1, len(seq)):

                    in_seq = seq[:i]
                    out_seq = seq[i]

                    in_seq = pad_sequences(
                        [in_seq],
                        maxlen=max_length
                    )[0]

                    out_seq = tf.keras.utils.to_categorical(
                        [out_seq],
                        num_classes=vocab_size
                    )[0]

                    X1.append(feature)
                    X2.append(in_seq)
                    y.append(out_seq)

                    if len(X1) == batch_size:

                        yield (
                            (np.array(X1), np.array(X2)),
                            np.array(y)
                        )

                        X1, X2, y = [], [], []
```

```python
gen = data_generator(train, 4)

(X1, X2), y = next(gen)

print(X1.shape)
print(X2.shape)
print(y.shape)
```

```python
# ============================================================
# CELL 10 — TRAINING SETUP (OPTIMIZED)
# ============================================================

from tensorflow.keras.callbacks import ModelCheckpoint, ReduceLROnPlateau, EarlyStopping

BATCH_SIZE = 64
EPOCHS = 25

# Count samples
def count_samples(image_list):
    total = 0
    for img in image_list:
        if img not in captions:
            continue
        for cap in captions[img]:
            seq = tokenizer.texts_to_sequences([cap])[0]
            total += (len(seq) - 1)
    return total

train_samples = count_samples(train)
val_samples   = count_samples(val)

train_steps = train_samples // BATCH_SIZE
val_steps   = val_samples // BATCH_SIZE

print("Train steps:", train_steps)
print("Val steps:", val_steps)


# Generators
train_gen = data_generator(train, BATCH_SIZE)
val_gen   = data_generator(val, BATCH_SIZE)


# Callbacks
checkpoint = ModelCheckpoint(
    "/content/best_model.keras",
    monitor="val_loss",
    save_best_only=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

print("✅ Training setup ready")
```

```python
# ============================================================
# CELL 11 — FINAL TRAINING
# ============================================================

history = model.fit(

    train_gen,

    steps_per_epoch=train_steps,

    epochs=25,

    validation_data=val_gen,

    validation_steps=val_steps,

    callbacks=[checkpoint, reduce_lr, early_stop],

    verbose=1
)

print("✅ Training completed")
```
